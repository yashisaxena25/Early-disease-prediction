```python
# ==============================
# DATA CLEANING
# ==============================

# Make a copy of the original dataset
df_clean = df.copy()

# 1. Remove unnecessary spaces from column names
df_clean.columns = df_clean.columns.str.strip()

# 2. Clean disease names
df_clean["Name"] = (
    df_clean["Name"]
    .str.strip()
)

# 3. Clean symptoms
df_clean["Symptoms_clean"] = (
    df_clean["Symptoms"]
    .str.lower()
    .str.strip()
)

# 4. Remove extra spaces around commas
df_clean["Symptoms_clean"] = (
    df_clean["Symptoms_clean"]
    .str.replace(r"\s*,\s*", ", ", regex=True)
)

# 5. Clean treatment text
df_clean["Treatments"] = (
    df_clean["Treatments"]
    .fillna("")
    .str.strip()
)

# 6. Check missing values
print("Missing values:")
print(df_clean.isnull().sum())

# 7. Check duplicate rows
print("\nDuplicate rows:", df_clean.duplicated().sum())

# 8. Basic dataset information
print("\nDataset shape:", df_clean.shape)

# 9. Number of unique diseases
print("Unique diseases:", df_clean["Name"].nunique())

# 10. Number of unique symptoms
unique_symptoms = (
    df_clean["Symptoms_clean"]
    .str.split(",")
    .explode()
    .str.strip()
    .nunique()
)

print("Unique symptoms:", unique_symptoms)

# 11. Display cleaned data
df_clean.head()
```

```python
from sklearn.feature_extraction.text import TfidfVectorizer

vectorizer = TfidfVectorizer(
    lowercase=True,
    ngram_range=(1, 2),
    min_df=1
)

X = vectorizer.fit_transform(df_clean["Symptoms_clean"])

y = df_clean["Name"]

print("Feature matrix shape:", X.shape)
print("Target shape:", y.shape)
```

```python

disease_counts = df_clean["Name"].value_counts()

print("Total diseases:", disease_counts.nunique())
print("\nDisease frequency:")
print(disease_counts.value_counts().sort_index())

print("\nDiseases with only 1 record:", (disease_counts == 1).sum())
print("Diseases with 2 records:", (disease_counts == 2).sum())
print("Diseases with 3+ records:", (disease_counts >= 3).sum())
```

```python
from sklearn.metrics.pairwise import cosine_similarity

similarity_matrix = cosine_similarity(X)

print("Similarity matrix shape:", similarity_matrix.shape)
```

```python


def predict_disease(user_symptoms, top_n=5):

    user_symptoms = user_symptoms.lower().strip()

    user_vector = vectorizer.transform([user_symptoms])

    similarities = cosine_similarity(user_vector, X).flatten()


    top_indices = similarities.argsort()[-top_n:][::-1]

    results = df_clean.iloc[top_indices][
        ["Name", "Symptoms", "Treatments"]
    ].copy()

    results["Similarity"] = similarities[top_indices]

    return results
```

```python
predict_disease(
    "fever, cough, fatigue, headache",
    top_n=5
)
```

```python


def predict_disease(user_symptoms, top_n=5):

    user_symptoms = user_symptoms.lower().strip()

    user_vector = vectorizer.transform([user_symptoms])

    similarities = cosine_similarity(user_vector, X).flatten()

    results = df_clean[["Name", "Symptoms", "Treatments"]].copy()
    results["Similarity"] = similarities

    results = results.sort_values("Similarity", ascending=False)
    results = results.drop_duplicates(subset="Name")

    
    return results.head(top_n).reset_index(drop=True)
```

```python
predict_disease(
    "fever, cough, fatigue, headache",
    top_n=5
)
```

```python
# ==============================
# EVALUATE SIMILARITY SYSTEM
# ==============================

correct = 0
total = len(df_clean)

for i in range(total):

    # Take one record's symptoms
    test_symptoms = df_clean.iloc[i]["Symptoms_clean"]
    actual_disease = df_clean.iloc[i]["Name"]

    # Convert symptoms to TF-IDF
    test_vector = vectorizer.transform([test_symptoms])

    # Calculate similarity with all records
    similarities = cosine_similarity(test_vector, X).flatten()

    # Ignore the record itself
    similarities[i] = -1

    # Find the most similar record
    predicted_index = similarities.argmax()
    predicted_disease = df_clean.iloc[predicted_index]["Name"]

    if predicted_disease == actual_disease:
        correct += 1

accuracy = correct / total

print("Correct predictions:", correct)
print("Total predictions:", total)
print("Similarity accuracy:", round(accuracy * 100, 2), "%")
```
