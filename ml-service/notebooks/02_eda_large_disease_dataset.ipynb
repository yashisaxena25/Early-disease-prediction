```python
import pandas as pd

path = "../../disease_symptoms/Final_Augmented_dataset_Diseases_and_Symptoms.csv"

df_large = pd.read_csv(path)

print("Dataset loaded!")
print("Rows:", df_large.shape[0])
print("Columns:", df_large.shape[1])
```

```python
# ==============================
# DUPLICATE ANALYSIS
# ==============================

duplicate_rows = df_large[df_large.duplicated(keep=False)]

print("Total duplicate rows:", len(duplicate_rows))

print("\nDuplicate rows by disease:")
print(duplicate_rows["diseases"].value_counts().head(20))

# Check duplicate symptom patterns
symptom_columns = df_large.columns[1:]

duplicate_symptoms = df_large.duplicated(
    subset=symptom_columns,
    keep=False
)

print("\nRows with duplicate symptom patterns:", duplicate_symptoms.sum())

# Check whether identical symptoms have different diseases
symptom_groups = (
    df_large
    .groupby(list(symptom_columns), sort=False)["diseases"]
    .nunique()
)

print(
    "\nSymptom patterns associated with multiple diseases:",
    (symptom_groups > 1).sum()
)
```

```python
# ==============================
# UNIQUE SYMPTOM PATTERNS
# ==============================

symptom_columns = df_large.columns[1:]

unique_patterns = df_large.drop_duplicates(
    subset=symptom_columns
)

print("Original rows:", len(df_large))
print("Unique symptom patterns:", len(unique_patterns))

print(
    "Rows removed if keeping one copy of each pattern:",
    len(df_large) - len(unique_patterns)
)

# How many patterns have exactly one disease?
pattern_disease_counts = (
    df_large
    .groupby(list(symptom_columns), sort=False)["diseases"]
    .nunique()
)

print(
    "\nPatterns belonging to exactly ONE disease:",
    (pattern_disease_counts == 1).sum()
)

print(
    "Patterns belonging to MULTIPLE diseases:",
    (pattern_disease_counts > 1).sum()
)
```

```python
# ==============================
# CREATE CLEAN DATASET
# ==============================

import numpy as np
import pandas as pd

symptom_columns = df_large.columns[1:]

# Convert symptoms to compact int8 representation
X_symptoms = df_large[symptom_columns].to_numpy(dtype=np.int8)

# Create a unique hash for each symptom pattern
pattern_hash = pd.util.hash_pandas_object(
    df_large[symptom_columns],
    index=False
)

# Count how many different diseases each pattern has
pattern_info = (
    pd.DataFrame({
        "pattern": pattern_hash,
        "disease": df_large["diseases"].values
    })
    .groupby("pattern")["disease"]
    .nunique()
)

# Patterns associated with exactly one disease
valid_patterns = pattern_info[pattern_info == 1].index

# Keep only unambiguous patterns
mask = pattern_hash.isin(valid_patterns)

clean_df = (
    df_large.loc[mask]
    .drop_duplicates(subset=symptom_columns)
    .reset_index(drop=True)
)

print("Original rows:", len(df_large))
print("Clean dataset rows:", len(clean_df))
print("Columns:", clean_df.shape[1])

print("\nNumber of diseases:", clean_df["diseases"].nunique())

print("\nDisease distribution:")
print(clean_df["diseases"].value_counts().describe())

print("\nDiseases with fewer than 10 samples:")
print(
    clean_df["diseases"]
    .value_counts()
    .loc[lambda x: x < 10]
)
```

```python
# ==============================
# FINAL ML DATASET
# ==============================

MIN_SAMPLES = 10

disease_counts = clean_df["diseases"].value_counts()

valid_diseases = disease_counts[
    disease_counts >= MIN_SAMPLES
].index

final_df = clean_df[
    clean_df["diseases"].isin(valid_diseases)
].reset_index(drop=True)

print("Final dataset shape:", final_df.shape)

print("Number of diseases:", final_df["diseases"].nunique())

print("\nDisease distribution:")
print(final_df["diseases"].value_counts().describe())

print("\nSmallest disease class:")
print(final_df["diseases"].value_counts().tail(10))
```

```python
# ==============================
# PREPARE FEATURES AND TARGET
# ==============================

from sklearn.model_selection import train_test_split

# Symptoms
X = final_df.drop(columns=["diseases"])

# Disease labels
y = final_df["diseases"]

print("X shape:", X.shape)
print("y shape:", y.shape)

print("\nNumber of features:", X.shape[1])
print("Number of diseases:", y.nunique())
```

```python
# ==============================
# TRAIN / TEST SPLIT
# ==============================

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

print("Training samples:", X_train.shape[0])
print("Testing samples:", X_test.shape[0])
```

```python
# ==============================
# TRAIN MODEL
# ==============================

from sklearn.svm import LinearSVC
from sklearn.metrics import accuracy_score, f1_score, classification_report
import time

model = LinearSVC(
    C=1.0,
    max_iter=3000
)

start = time.time()

model.fit(X_train, y_train)

print("Training completed!")
print("Time:", round(time.time() - start, 2), "seconds")
```

```python
# ==============================
# EVALUATE MODEL
# ==============================

y_pred = model.predict(X_test)

accuracy = accuracy_score(y_test, y_pred)
macro_f1 = f1_score(y_test, y_pred, average="macro")
weighted_f1 = f1_score(y_test, y_pred, average="weighted")

print("Accuracy:", round(accuracy, 4))
print("Macro F1:", round(macro_f1, 4))
print("Weighted F1:", round(weighted_f1, 4))
```

```python
import joblib
import os

os.makedirs("../models", exist_ok=True)

joblib.dump(model, "../models/disease_model.joblib")

joblib.dump(
    list(X.columns),
    "../models/symptom_columns.joblib"
)

print("Model saved successfully!")
```

```python
# ==============================
# DISEASE PREDICTION
# ==============================

def predict_disease(symptoms):
    """
    symptoms = list of symptom column names
    Example:
    ["fever", "headache", "cough"]
    """

    input_data = pd.DataFrame(
        0,
        index=[0],
        columns=X.columns
    )

    for symptom in symptoms:
        if symptom in input_data.columns:
            input_data.loc[0, symptom] = 1

    prediction = model.predict(input_data)[0]

    return prediction


# Test
result = predict_disease([
    "fever",
    "headache",
    "cough"
])

print("Predicted disease:", result)
```
